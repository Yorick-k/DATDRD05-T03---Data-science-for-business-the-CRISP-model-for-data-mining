# Phase 1 – Business Understanding

## 1.1 Business Problem

The organisation wants to identify employees who may leave
within the next six months.

Employee attrition can affect staffing continuity and may
require the organisation to recruit and train replacements.

The goal is to support HR in identifying employees who may
benefit from a voluntary, human-led retention review.

The model is used as decision support.
It does not make automatic personnel decisions.
HR always reviews the prediction before taking action.


## 1.2 Stakeholder and Business Decision

**Stakeholder:** HR department / HR manager

**Operational decision:** Identify employees with an elevated
risk of leaving who may be considered for a voluntary
retention review.

**Prediction moment:** Employee survey date

**Prediction horizon:** Next six months

**Target variable:** `attrition_next_6m`

Only information that is available at the prediction moment
may be used by the model.


## 1.3 Success Criteria

The model will be compared with a majority-class baseline.

The main evaluation measures are:

- Precision
- Recall
- False positives and false negatives
- Performance across departments
- Fairness between departments

Recall shows how many employees who actually leave are
correctly identified by the model.

Precision shows how many employees predicted to leave
actually leave.

A false negative means that an employee who eventually
leaves is not identified for a retention review.

A false positive means that HR may spend time reviewing
an employee who would not have left.

Performance should also be compared across departments
to identify possible fairness concerns.

HR must review the model output before any action is taken.


## 1.4 Limitations of the Business Goal

No specific financial saving or reduction in employee
attrition is claimed.

The available data does not provide verified cost
assumptions to reliably calculate the financial impact.

The model predicts the risk of attrition, but this does not
prove that a retention review will cause an employee to stay.


## 1.5 Business Objective

The final business objective is to support HR in identifying
employees who may be at increased risk of leaving within
the next six months.

The classification model will support voluntary,
human-led retention reviews while considering prediction
quality, fairness and human oversight.


## Phase 1 Conclusion

The model will support HR decision-making rather than
replace human judgement.

The next phase examines whether the available employee
data is suitable and reliable enough to support this
business objective.
